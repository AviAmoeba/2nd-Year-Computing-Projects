**Question 1.** Implement the function Gauss(A,b) that takes as inputs an $n\times n$ matrix $A$ and a vector $b$ of length $n$ and performs the Gaussian elimination algorithm on the augmented matrix $(A,b)$. Test the function by printing the result of Gauss(A_1,b).

```python
# Use this cell to implement the function Gauss(A,b) for Gaussian elimination algorithm

def Gauss(A, b):
    
    A = A.astype(float)
    b = b.astype(float)
    n = A.shape[0]
    
    for l in range(n):
        for i in range(l+1, n):
            factor = A[i, l] / A[l, l]

            for j in range(l, n):
                A[i, j] -= factor * A[l, j]

            b[i] -= factor * b[l]
    
    return A, b

eli_matrix, b_eli = Gauss(A1, b)
print(eli_matrix)
```

**Question 2.** Implement two functions: the backwards substitution BS(A,b) and the forwards substitution FS(A,b) that take as inputs the upper triangular or the lower triangular matrix $A$, respectively, and the vector $b$. Solve $A_1x=b$ and print your solution $x$.

```python
# Use this cell to implement functions BS(A,b) and FS(A,b) and solve A_1x=b

def BS(U, b):
    U = U.astype(float)
    b = b.astype(float)
    n = U.shape[0]
    
    x = np.zeros(n)
    
    for i in range(n-1, -1, -1):
        x[i] = (b[i] - np.dot(U[i, i+1:], x[i+1:])) / U[i, i]
    
    return x


def FS(L, b):
    L = L.astype(float)
    b = b.astype(float)
    n = L.shape[0]
    
    x = np.zeros(n)
    
    for i in range(n):
        x[i] = (b[i] - np.dot(L[i, :i], x[:i])) / L[i, i]
    
    return x

x = BS(eli_matrix, b_eli)
print(x)
```

**Question 3.** Implement the function GaussJordan(A,b) for the Gauss-Jordan elimination method. Solve the equation $A_1x=b$ using the Gauss-Jordan elimination method and print your result.

```python
# Use this cell to implement the function GaussJordan(A,b) and to solve A_1x=b using the Gauss-Jordan elimination algorithm
def GaussJordan(U, b):

    U = U.astype(float)
    b = b.astype(float)
    n = U.shape[0]

    for i in range(n-1, -1, -1):

        pivot = U[i, i]
        U[i, :] /= pivot
        b[i] /= pivot

        for j in range(i-1, -1, -1):
            factor = U[j, i]
            U[j, :] -= factor * U[i, :]
            b[j] -= factor * b[i]

    return b

y, z = Gauss(A1, b)
x = GaussJordan(y, z)
print(x)
```

**Question 4.** Write new functions GaussRounded(A,b,r) and BSRounded(A,b,r) such that each floating number in calculations within these functions is rounded to $r$ decimal places after comma. For $r=3, 4, 5, 6, 7$ solve $A_1x=b$ and $A_2x=b$ and print your results.

```python
# Use this cell to implement GaussRounded(A,b,r) and BSRounded(A,b,r) and to solve A_1x=b and A_2x=b for r=3,4,5,6,7

def GaussRounded(A, b, r):
    A = A.astype(float)
    b = b.astype(float)
    n = A.shape[0]
    
    for l in range(n):
        for i in range(l+1, n):
            factor = round(A[i, l] / A[l, l], r)  
            for j in range(l, n):
                A[i, j] = round(A[i, j] - round(factor * A[l, j], r), r)  
            b[i] = round(b[i] - round(factor * b[l], r), r)
    
    return A, b

# Back-substitution with rounding
def BSRounded(U, b, r):
    U = U.astype(float)
    b = b.astype(float)
    n = U.shape[0]
    
    x = np.zeros(n)
    
    for i in range(n-1, -1, -1):
        s = 0
        for j in range(i+1, n):
            s = round(s + round(U[i, j] * x[j], r), r) 
        x[i] = round((b[i] - s) / U[i, i], r)
    
    return x

print()
print("Using Rounded Gaussian Elimination on A1")
for i in range(3, 8):
    A_val, b_val = GaussRounded(A1, b, i)
    x_val = BSRounded(A_val, b_val, i)
    print(x_val)

print()
print("Using Rounded Gaussian Elimination on A2")
for i in range(3, 8):
    A_val, b_val = GaussRounded(A2, b, i)
    x_val = BSRounded(A_val, b_val, i)
    print(x_val)

```

**Question 6.** Implement the function LU(A) that returns the LU factorisation of the matrix $A$ and solve $A_1x=b$ using the LU factorisation as well as forwards and backwards substitution.

```python
# Use this cell to implement LU(A) and to solve A_1x=b using LU factorisation

def LU(A):
    A = A.astype(float)
    n = A.shape[0]
    
    L = np.eye(n)      
    U = A.copy()
    
    for k in range(n):
        for i in range(k+1, n):
            factor = U[i, k] / U[k, k]
            L[i, k] = factor
            
            for j in range(k, n):
                U[i, j] -= factor * U[k, j]
    
    return L, U

l, u = LU(A1)
y = FS(l, b)
x = BS(u, y)

print(x)
```
